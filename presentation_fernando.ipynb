## 5. Réponses aux 3 questions d'acculturation IA

### 5.1 Apprentissage supervisé vs non supervisé

L’apprentissage supervisé utilise des données avec des réponses déjà connues afin d’apprendre à faire une prédiction.
Par exemple, une entreprise peut entraîner un modèle à détecter les e-mails frauduleux à partir d’e-mails déjà classés.
L’apprentissage non supervisé cherche des groupes ou des structures dans des données sans réponse prédéfinie.
Par exemple, il peut servir à regrouper des clients selon leurs habitudes d’achat.
### 5.2 À quoi sert le transfer learning ?

Le transfer learning permet de partir d’un modèle déjà entraîné sur une grande quantité de données et de l’adapter à un besoin plus précis.
Cela permet de gagner du temps et d’utiliser moins de données pour l’entraînement.
Par exemple, un modèle de reconnaissance d’images peut être adapté pour détecter des défauts sur des produits industriels.

### 5.3 Pourquoi un LLM n'est pas toujours la bonne réponse ?

Un LLM peut être coûteux, lent et produire des réponses incorrectes ou difficiles à contrôler.
Pour un problème simple et déterministe, une règle métier, une requête SQL ou un modèle classique est souvent plus fiable.
Par exemple, utiliser un LLM pour calculer une remise selon un pourcentage fixe serait surdimensionné.
